Nesta questão, você irá calcular a diferença no número total de palavras entre textos negativos em português e inglês.

### Objetivo:

```python
def map2(x):
  # Conta as palavras em inglês e português para cada linha
  qtd_en = len(x['text_en'].split()) if x['text_en'] else 0
  qtd_pt = len(x['text_pt'].split()) if x['text_pt'] else 0
  return (x['sentiment'], (qtd_en, qtd_pt))
```

```python
def reduceByKey2(x, y):
  # Soma as contagens de palavras em inglês e português separadamente
  return (x[0] + y[0], x[1] + y[1])
```

## Aplicação do map/reduce e visualização do resultado

1. Aplicar o map/reduce no seu dataframe spark e realizar o collect() ao final
2. Selecionar os dados referentes aos textos negativos para realizar a subtração.
3. Realizar a subtração das contagens de palavras dos textos negativos para obter o resultado final

```python
# Aplica o map/reduce no RDD do dataframe spark
resultados_palavras = imdb_df.rdd.map(map2).reduceByKey(reduceByKey2).collect()

# Filtra os totais de palavras em inglês e português para o sentimento 'neg'
totais_neg = next(valores for chave, valores in resultados_palavras if chave == 'neg')
total_en, total_pt = totais_neg

diferenca = total_pt - total_en

print(f"RU: 5082456")
print(f"Total de palavras em Inglês (neg): {total_en}")
print(f"Total de palavras em Português (neg): {total_pt}")
print(f"Diferença (Português - Inglês): {diferenca}")
```
